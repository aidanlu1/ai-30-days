Change the hours in the first cell to your real week, then run **Kernel → Restart & Run All**.

```python
# 1. Subjects and hours studied this week
hours = {"CS": 6, "Econ": 4, "Writing": 3, "Math": 5}
hours
```

```python
# 2. Total hours
def total_hours(hours_by_subject):
    return sum(hours_by_subject.values())

total_hours(hours)
```

```python
# 3. Subject studied most
def most_studied(hours_by_subject):
    max_hours = 0
    best_subject = None
    for subject, h in hours_by_subject.items():
        if h > max_hours:
            max_hours = h
            best_subject = subject
    return best_subject

most_studied(hours)
```

```python
# 4. Bar chart: one # per hour
for subject, h in hours.items():
    print(subject, "#" * h)
```
